```python
import pandas as pd
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df1 = pd.read_csv("../outputs/tables/sample/repeated_kfold_scores.csv")
df2 = pd.read_csv("../outputs/tables/sample/spatial_block_cv_scores.csv")
combined = pd.concat([df1, df2])
```

```python
fig, ax = plt.subplots(figsize=(6, 5))

my_colors = {
    'repeated_stratified_kfold': '#1f77b4',
    'spatial_block_cv': '#ff7f0e'
}

sns.boxplot(
    x="cv_method", 
    y="accuracy", 
    hue="cv_method", 
    palette=my_colors, 
    legend=False, 
    data=combined, 
    ax=ax
)

ax.set_title("Model accuracy: repeated stratified k-fold vs. spatial block CV")
ax.set_xlabel("")
ax.set_ylabel("Accuracy")

plt.savefig("../outputs/figures/sample/compare_cv_methods_accuracy.png", bbox_inches="tight", dpi=300)
plt.show()
```
